# 08 — Final Case Study Reporting

This notebook assembles the completed case study into a standalone HTML report aligned with the assignment rubric. It reads the saved outputs from Steps 01–07, creates a compact EDA summary figure, embeds every figure directly into the HTML, and performs no model fitting, selection, threshold adjustment, or test reevaluation.

```python
from pathlib import Path
import sys

project_root = Path.cwd()
if not (project_root / "data" / "raw").exists():
    project_root = project_root.parent
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))
from src.reporting.build_report import build_final_report
print("Project root:", project_root)
```

## 1. Build the Standalone HTML Report

The generated file embeds images as base64 data, so the submitted HTML does not depend on separate figure files.

```python
report_path = build_final_report(project_root)
print("Final report created:", report_path.relative_to(project_root))
print(f"File size: {report_path.stat().st_size / 1024**2:.2f} MiB")
```

## 2. Validate Submission Artifact

```python
report_html = report_path.read_text(encoding="utf-8")
required_sections = ["Executive Summary", "Business Problem and Objective", "Data Overview and Exploratory Analysis", "Preprocessing and Leakage Controls", "Feature Engineering", "Baseline Model Comparison", "Hyperparameter Tuning and Threshold Selection", "Final Test Performance", "Feature Importance", "Actionable Insights and Recommendations", "Limitations and Responsible Use", "Conclusion"]
validation = {"valid_html_shell": report_html.lstrip().startswith("<!doctype html>") and report_html.rstrip().endswith("</html>"), "all_required_sections_present": all(section in report_html for section in required_sections), "figures_embedded": report_html.count("data:image/png;base64,") >= 7, "under_100_mib": report_path.stat().st_size < 100 * 1024**2, "contains_no_jupyter_error_output": "Traceback (most recent call last)" not in report_html}
for check, passed in validation.items():
    print(f"{check}: {passed}")
if not all(validation.values()):
    raise RuntimeError("Final HTML report validation failed.")
```

## 3. Report Contents

```python
print("The final report includes:")
for section in required_sections:
    print(f"- {section}")
```
